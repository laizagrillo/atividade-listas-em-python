```python
#operações
# + soma - subtrai * multiplica / divisão completa % resto da divisão (módulo) // divisão sem decimal ** potencia

n1 = int(input("Digite um número: "))
n2 = int(input("Digite outro número: "))

resto = n1 % n2

print(f"O resto da divisão de {n1} por {n2} é {resto}")

#potencia
base = int(input("Digite a base: "))
expoente = int(input("Digite o expoente: "))

potencia = base ** expoente

print(f"{base} elevado a {expoente} é igual a {potencia}")

# raiz
ind = int(input("Digite o índice: "))
numero = int(input("Digite o número que vc quer a raiz: "))
raiz = numero ** (1/ind)

print(f"A raiz de {numero} é {raiz}")
```

```python
#CRIANDO LISTAS INICIAIS

produtos = ["Camiseta Básica", "Calça Jeans", "Jaqueta Corta-Vento"]
precos = [49.90, 129.90, 219.90]

#CADASTRANDO NOVOS ITENS(Leitura e append)

resposta = input("Deseja cadastrar novo produto?")
if resposta.lower() == "sim":
  novo_produto = input("Insira o nome do produto: ")
  novo_preco = float(input("Informe o valor do produto (com ponto): "))

produtos.append(novo_produto)
precos.append(novo_preco)

for i,j in zip(produtos, precos):
  print(f"O produto {i} custa R$ {j:.2f}")

#REMOVER ÚLTIMO ITEM

if len(produtos) > 0:
    produto_removido = produtos.pop()
    preco_removido = precos.pop()
    print(f"Produto removido: {produto_removido} (Preço: {preco_removido:.2f})")
else:
    print("Estoque vazio!")

```

```python
pergunta1 = input("Deseja remover um produto? ")
if pergunta1.lower() == "sim":
  produto_para_remover = input("Qual produto deseja remover?")

  if produto_para_remover in produtos:
    posicao_do_produto = produtos.index(produto_para_remover)
    pr_removido = produtos.pop(posicao_do_produto)
    pre_removido = precos.pop(posicao_do_produto)
    print(f"O produto {pr_removido} foi removido e custava R$ {pre_removido:.2f}")
  else:
    print(f"O produto '{produto_para_remover}' não foi encontrado no estoque.")
else:
  print("Interação finalizada")
```
